```python
#problem41: Creating a Distance Matrix
#p-distance 를 구하라
#p-distance:  두 DNA 문자열의 차이를 비율로 나타낸 유전 거리
#dp(s1, s2) = 다른 base 수 / 전체 length

data = {}
with open("p41_sample.fasta") as handle:
    for line in handle:
        line = line.strip()
        if line.startswith(">"):
            id = line
            data[id] = ""
        else:
            data[id] = line

def p_distance(s1, s2):
    full_length = len(s1) #어차피 길이 다 같아서 하나만 잼
    diff = 0
    for base1, base2 in zip(s1, s2):
        if base1 != base2:
            diff += 1
    result = diff / full_length
    return result

#행렬 만들기
def build_p_distance_matrix(data = dict):
    global p_distance
    ids = list(data.keys())
    n = len(ids)
    matrix = []

    for i in range(n):
        row = []
        for j in range(n):
            s1 = data[ids[i]]
            s2 = data[ids[j]]
            if i == j:
                row.append(0.0) #자기 자신은 0
            else:
                row.append(p_distance(s1, s2))
        matrix.append(row)

    return matrix


matrix = build_p_distance_matrix(data)

for row in matrix:
    print(' '.join(f"{val:.3f}" for val in row))
    
```
